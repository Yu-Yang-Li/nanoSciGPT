# 用 32 个标签估计模拟链的首尾距

先在仓库根目录用内核对应的 Python 运行 `python -m pip install -e .`。依次运行下面三格。两组确认样本各有 480 条链，监督标签和模型结构在三档间固定。

```python
from nanoscigpt.tasks.structure_scale import chain_panel
visible, length = chain_panel(1, [264130, 2])
print("可见块、每块的距离数", tuple(visible.shape[1:]))
print("第一块中的前四个距离", visible[0, 0, :4].tolist())
print("监督阶段才使用的首尾距", float(length[0]))
```

```python
import csv
import io
import subprocess
import sys

run = subprocess.run([sys.executable, "-m", "nanoscigpt.tasks.structure_scale"], capture_output=True, text=True, check=True)
rows = list(csv.DictReader(io.StringIO(run.stdout)))
assert [int(r["unlabeled_objects"]) for r in rows] == [200, 800, 3200]
print(run.stdout)
```

```python
for group in ["C_MAE", "D_MAE"]:
    first, middle, last = [float(row[group]) for row in rows]
    print(f"{group}：200 → 800 的误差变化 {middle - first:+.3f}，800 → 3,200 的变化 {last - middle:+.3f}")
```

固定种子 53 的三档曲线可与五种子均值对照。误差变化为负表示这一段改善。完整四档和五种子结果见[十一方向规模曲线](README.md)。
